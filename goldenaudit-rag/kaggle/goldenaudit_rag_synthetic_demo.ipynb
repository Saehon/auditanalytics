# GoldenAudit-RAG: Kaggle-ready synthetic BM25 pilot
**Not a validated ISACA exam model, LLM or golden-answer benchmark.** This stand-alone notebook retrieves only synthetic audit-control evidence; no API keys or private data are needed.

```python
import re, math
from collections import Counter
docs = [
 {'id':'SYN-IS-01','text':'When employee access is revoked, retain approval and review evidence.'},
 {'id':'SYN-AI-02','text':'High impact AI deployment requires model version, evaluation and human approval.'},
 {'id':'SYN-ICFR-03','text':'Missing reconciliation requires further audit evidence and a deficiency evaluation.'},
 {'id':'SYN-ESG-04','text':'ESG assurance needs the standard effective date, reporting scope and traceable evidence.'},
 {'id':'SYN-CAM-05','text':'A critical audit matter is not evidence by itself of misstatement or fraud.'},
]
def tok(s): return re.findall(r'[a-z0-9]+',s.lower())
def bm25(query,k=3):
    seq=[tok(x['text']) for x in docs]; avg=sum(map(len,seq))/len(seq)
    scored=[]
    for i,d in enumerate(docs):
        ct=Counter(seq[i]); value=0.0
        for word in set(tok(query)):
            df=sum(word in x for x in seq)
            idf=math.log(1+(len(seq)-df+0.5)/(df+0.5))
            f=ct[word]
            if f: value+=idf*f*2.5/(f+1.5*(0.25+0.75*len(seq[i])/avg))
        scored.append((value,d['id'],d['text']))
    return sorted(scored,key=lambda x:(-x[0],x[1]))[:k]
question='What is required before high impact AI deployment?'
for score,doc_id,passage in bm25(question): print(f'{doc_id}: {score:.3f} | {passage}')
assert bm25('AI deployment human approval')[0][1]=='SYN-AI-02'
```
